**#L3M1S1: List to List comprehension and various methods**

```python
# Level 1
texts = ["A-1","b_2","C!3"]
clean = [''.join(ch for ch in s if ch.isalnum()).lower() for s in texts]
tmp = clean
ans1 = tmp
#ans1 = ['a1', 'b2', 'c3']


# Level 2
texts = ["A-1","b_2","C!3"]
clean = [''.join(ch for ch in s if ch.isalnum()).upper() for s in texts]
tmp = clean
ans2 = tmp
#ans2 = ['A1', 'B2', 'C3']


# Level 3
texts = ["A-1","b_2","C!3"]
clean = [''.join(ch for ch in s if ch.isdigit()).lower() for s in texts]
tmp = clean
ans3 = tmp
#ans3 = ['1', '2', '3']
```

**#L3M3S1:  Two types of sort techniques (function & method)**

```python
# Level 1
data = [3,1,2]
data.sort()
ans1 = (data == sorted(data))
#ans1 = True


# Level 2
data = [3,1,2]
safe_sorted = sorted(data)
ret = data.sort()
ans2 = (data, safe_sorted, ret is None)
#ans2 = ([1, 2, 3], [1, 2, 3], True)


# Level 3
data = [3,1,2]
safe_sorted = sorted(data)
ret = data.sort()
ans3 = (id(safe_sorted) != id(data), ret is None)
#ans3 = (True, True)
```

```python
ans1,ans2,ans3
```

**#L3M4S1: Pandas's date-type data (2)**

```python
# Level 1
import pandas as pd
df = pd.DataFrame({"date":["2023/04/01","2023/04/02"], "x":[1,2]})
df["date"] = pd.to_datetime(df["date"])
df = df[["x","date"]]
tmp = (df.columns.tolist(), df["date"].dt.day.tolist())
ans1 = tmp
#ans1 = (['x','date'], [1, 2])


# Level 2
import pandas as pd
df = pd.DataFrame({"date":["2023/04/01","2023/04/02"], "x":[1,2]})
df["date"] = pd.to_datetime(df["date"])
df = df[["x","date"]]
tmp = (df.columns.tolist(), df["date"].dt.month.tolist())
ans2 = tmp
#ans2 = (['x','date'], [4, 4])


# Level 3
import pandas as pd
df = pd.DataFrame({"date":["2023/04/01","2023/04/02"], "x":[1,2]})
df["date"] = pd.to_datetime(df["date"])
df = df[["x","date"]]
tmp = (df.columns.tolist(), df["date"].dt.year.tolist())
ans3 = tmp
#ans3 = (['x','date'], [2023, 2023])
```

```python
ans1,ans2,ans3
```

**#L3M6S1: Pandas's date-type data structure (4)**

```python
# Level 1
import pandas as pd
df = pd.DataFrame({
    "flag":[True, False, True],
    "name":["u00","u01","u02"],
    "date": pd.to_datetime(["2023/04/01","2023/04/02","2023/04/03"]),
    "value":[10,20,30]
})
df2 = df.copy()
ans1 = (id(df["date"]) != id(df2["date"]), df["value"].sum() == df2["value"].sum())
#ans1 =  (True, True)


# Level 2
import pandas as pd
df = pd.DataFrame({
    "flag":[True, False, True],
    "name":["u00","u01","u02"],
    "date": pd.to_datetime(["2023/04/01","2023/04/02","2023/04/03"]),
    "value":[10,20,30]
})
ans2 = (df.dtypes["date"].name, df["value"].sum())
#ans2 = ('datetime64[ns]', 60)


# Level 3
import pandas as pd
df = pd.DataFrame({
    "flag":[True, False, True],
    "name":["u00","u01","u02"],
    "date": pd.to_datetime(["2023/04/01","2023/04/02","2023/04/03"]),
    "value":[10,20,30]
})
ans3 = (isinstance(df.loc[0, "date"], pd.Timestamp), df["value"].sum() == 60)
#ans3 = (True, True)
```

```python
ans1,ans2,ans3
```

**#L3M7S1: Data analysis sample**

```python
# Level 1
import numpy as np
items  = np.array([1000,1001,1001,1002])
stores = np.array([1,2,2,3])
qty    = np.array([5,7,3,2])
grid = np.zeros((3,3), dtype=int)
for r, c, q in zip(items-1000, stores-1, qty):
    grid[r, c] += int(q)
tmp = (grid.shape, int(grid.sum()), int(grid[1,1]))
ans1 = tmp
#ans1 = ((3, 3), 17, 10)


# Level 2
import numpy as np
items  = np.array([1000,1001,1001,1002])
stores = np.array([1,2,2,3])
qty    = np.array([6,1,3,2])
grid = np.zeros((3,3), dtype=int)
for r, c, q in zip(items-1000, stores-1, qty):
    grid[r, c] += int(q)
tmp = (grid.shape, int(grid.sum()), int(grid[1,1]))
ans2 = tmp
#ans2 =  ((3, 3), 12, 4)


# Level 3
import numpy as np
items  = np.array([1000,1001,1001,1002,1002])
stores = np.array([1,2,2,3,1])
qty    = np.array([1,8,4,2,10])
grid = np.zeros((3,3), dtype=int)
for r, c, q in zip(items-1000, stores-1, qty):
    grid[r, c] += int(q)
tmp = (grid.shape, int(grid.sum()), int(grid[1,1]))
ans3 = tmp
#ans3 = ((3, 3), 25, 12)
```
